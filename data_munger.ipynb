```python
import pandas as pd
```

```python
def preprocess_data(file_path):
    df = pd.read_csv(file_path)
    df_new = df[df['product'] == 'pink morsel'].copy()
    
    # Clean price column by removing $ and convert to float
    df_new['price'] = df_new['price'].replace(r'[\$,]', '', regex=True).astype(float)
    
    # Calculate sales
    df_new['sales'] = df_new['price'] * df_new['quantity']
    df_new['sales'] = df_new['sales'].apply(lambda x: f"${x:.2f}")
    
    # Drop unnecessary columns
    df_new = df_new.drop(['product', 'price', 'quantity'], axis=1)
    
    # Rearrange the columns
    df_new = df_new[['sales', 'date', 'region']]
    
    return df_new
```

```python
df1= preprocess_data('data/daily_sales_data_0.csv')
df2= preprocess_data('data/daily_sales_data_1.csv')
df3= preprocess_data('data/daily_sales_data_2.csv')
```

```python
df_merged = pd.concat([df1, df2, df3], ignore_index=True)
```
